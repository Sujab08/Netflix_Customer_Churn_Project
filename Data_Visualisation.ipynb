```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.hist(df['last_login_days'],edgecolor="black")
plt.title('Last day login')
plt.grid(axis='y')
plt.show()
```

```python
plt.hist(df['age'],bins=[0,10,20,40,60,80,100],edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(axis='y',color='red')
plt.show()
```

```python
df['favorite_genre'].value_counts().index
```

```python
plt.figure(figsize=(10,6))
plt.bar(x=df['favorite_genre'].value_counts().index,height=df['favorite_genre'].value_counts().values)
plt.xlabel("Movie Type")
plt.ylabel("counts")
plt.grid(axis='y')
plt.show()
```

```python
payment=df['payment_method'].value_counts()
plt.figure(figsize=(7,4))
plt.bar(x=payment.index,height=payment.values,color='skyblue')
plt.xlabel('payment type')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.show()
```

```python
df['churned'].value_counts()
```

```python
pd.crosstab(df["gender"], df["churned"])
```

```python

churn_gender = pd.crosstab(df["gender"], df["churned"])
plt.figure(figsize=(20,5))
churn_gender.plot(kind="bar")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.title("Churn by Gender")
plt.xticks(rotation=0)
plt.legend(title="Churned")
plt.grid(axis="y")
plt.show()
```

```python
df.groupby(["gender", "churned"]).size()
```

```python
pd.crosstab(df["gender"], df["region"])
```

```python

region_gender = pd.crosstab(df["gender"], df["region"])
plt.figure(figsize=(15,15))
region_gender.plot(kind="bar")
plt.xlabel("region")
plt.ylabel("Count")
plt.title("gender by region")
plt.xticks(rotation=0)
plt.legend(title="region")
plt.grid(axis="y")
plt.show()
```

```python
favorite_genre_gender = pd.crosstab(df["gender"], df["favorite_genre"])
plt.figure(figsize=(15,15))
favorite_genre_gender.plot(kind="bar")
plt.xlabel("favorite_genre")
plt.ylabel("Count")
plt.title("gender by favorite_genre")
plt.xticks(rotation=0)
plt.legend(title="favorite_genre")
plt.grid(axis="y")
plt.show()
```

```python
region_churned = pd.crosstab(df["churned"], df["region"])
plt.figure(figsize=(15,15))
region_churned.plot(kind="bar")
plt.xlabel("region")
plt.ylabel("Count")
plt.title("Churn by region")
plt.xticks(rotation=0)
plt.legend(title="Churned")
plt.grid(axis="y")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd
df=pd.read_csv(r"C:\Users\kumar\Downloads\netflix_customer_churn (1).csv")
```

```python
df.head(5)
```

```python
plt.hist(df['age'],bins=[0,10,20,40,60,80,100],edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(axis='y',color='red')
plt.show()
```

```python
gender=df['gender'].value_counts()
```

```python
gender.values
```

```python
plt.pie(gender.values,labels=gender.index,autopct="%1.1f%%")
plt.title("Gender distribution")
plt.show()
```

```python
sub=df['subscription_type'].value_counts()
```

```python
plt.pie(sub.values,labels=sub.index,autopct="%1.1f%%")
plt.title("Subscriber distribution")
plt.show()
```

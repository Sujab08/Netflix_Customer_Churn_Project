```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv(r"C:\Users\kumar\Simple_Project\cleaned_netflix_data.csv")
df.head()
```

```python
df.shape
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
df.columns
```

```python
X=df.drop(columns='churned')
y=df['churned']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=42,test_size=0.2)
```

```python
X_test.shape
```

```python
y_test.shape
```

# Logistic Regression

```python
lg=LogisticRegression()
lg.fit(X_train,y_train)
```

```python
lg.score(X_test,y_test),lg.score(X_train,y_train)
```

```python
y_pre=lg.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,f1_score,recall_score,accuracy_score,classification_report
```

```python
f1_score(y_test,y_pre)
```

```python
recall_score(y_test,y_pre)
```

```python
accuracy_score(y_test,y_pre)
```

```python
print(classification_report(y_test, y_pre))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
cm = confusion_matrix(y_test, y_pre)
sns.heatmap(cm, annot=True, fmt="d")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

# Naive bayes

```python
from sklearn.naive_bayes import GaussianNB,MultinomialNB,BernoulliNB
```

```python
gs=GaussianNB()
gs.fit(X_train,y_train)
```

```python
print(gs.score(X_test,y_test),gs.score(X_train,y_train))
```

```python
y_pre=gs.predict(X_test)
```

```python
cm = confusion_matrix(y_test, y_pre)
sns.heatmap(cm, annot=True, fmt="d")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

```python
print("recall:",recall_score(y_test,y_pre), "  accuracy:",accuracy_score(y_test,y_pre),"  f1 score:",f1_score(y_test,y_pre))
```

```python
print(classification_report(y_test,y_pre))
```

# KNN

```python
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier(n_neighbors=10)
knn.fit(X_train,y_train)
```

```python
knn.score(X_test,y_test), knn.score(X_train,y_train)
```

```python
y_pre=knn.predict(X_test)
```

```python
print("recall:",recall_score(y_test,y_pre), "  accuracy:",accuracy_score(y_test,y_pre),"  f1 score:",f1_score(y_test,y_pre))
```

```python
cm = confusion_matrix(y_test, y_pre)
sns.heatmap(cm, annot=True, fmt="d")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

# Decision tree

```python
from sklearn.tree import DecisionTreeClassifier
dc=DecisionTreeClassifier()
dc.fit(X_train,y_train)
```

```python
dc.score(X_test,y_test),dc.score(X_train,y_train)
```

```python
for i in range (1,21):
    dc=DecisionTreeClassifier(max_depth=i)
    dc.fit(X_train,y_train)
    print(dc.score(X_test,y_test),dc.score(X_train,y_train),  i)
```

```python
y_pre=dc.predict(X_test)
```

```python
cm = confusion_matrix(y_test, y_pre)
sns.heatmap(cm, annot=True, fmt="d")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

```python
print(classification_report(y_test,y_pre))
```
